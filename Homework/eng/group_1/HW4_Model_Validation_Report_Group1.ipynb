### 2.Data processing
The project uses sklearn Pipelines for preprocessing, including separate pipelines for categorical and numerical features. Missing values are imputed and categorical variables are one-hot encoded. Feature engineering steps (e.g., creation of new variables) are encapsulated in a function. This improves consistency and reduces risks of accidental information leakage.

The preprocessing includes robust text mining using regular expressions to extract numerical features from unstructured text fields such as "house layout" and "property fees." Floor information is processed meticulously, extracting total floors while preserving categorical information like "high/middle/low" floors. Interaction terms and quadratic features are explicitly created to enhance feature expressiveness for linear models.

### 3.Sample choice
Train-test split logic separates Price and Rent datasets individually, which is appropriate. Preprocessing is wrapped inside pipelines, ensuring the transformers are fit only on the training data when performing cross-validation or prediction.

No explicit leakage was detected in the code structure. Feature creation occurs before splitting, but only uses raw input features without referencing target variables, so this does not constitute leakage. The training and test sets are pre-divided (Xp_train/Xp_test, Xr_train/Xr_test), with all preprocessing and feature engineering performed exclusively on the training set before transformation on the test set through Pipeline processing.

### 4. Suggestions
(1) To ensure better backward compatibility, consider adding version checks for scikit-learn dependencies and include a requirements.txt file with version specifications for key packages like scikit-learn. Furthermore, enhancing the documentation by adding more detailed docstrings for custom functions and pipeline steps would significantly improve code clarity and maintainability.

(2) Performing skewness transformation on the area variable helps make the numerical distribution closer to normal, which enhances model stability and reduces the impact of extreme values.

(3) Consider replacing SGDRegressor with direct implementations of Ridge(), Lasso(), or LinearRegression() from sklearn, as analytical solutions typically provide more stable baselines without requiring learning rate adjustments.

(4) For regularization parameters, the poor performance of Lasso and ElasticNet suggests potential underfitting due to excessive penalty. Recommend trying smaller alpha values (e.g., 1e-4, 1e-5) in GridSearch.

(5) Enhance geographical information utilization beyond simple longitude-latitude interaction terms. Consider implementing target encoding for plate or region features, or generating new regional characteristics using clustering algorithms based on coordinates.

(6) Although the pipeline structure is excellent, the main program section remains somewhat lengthy. The code can be decomposed into smaller functions, and similar data processing logic can be repeated multiple times using configuration file management parameters. A unified preprocessing pipeline can be created to avoid code duplication.
